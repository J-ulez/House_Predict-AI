```python
# Load the Saved Models

# Linear Regression model
with open(r"/content/drive/MyDrive/Software System Engineering/Coursework/Models/linear_regression_model.pkl", "rb") as file:
    linear_model = pickle.load(file)

# Random Forest model
with open(r"/content/drive/MyDrive/Software System Engineering/Coursework/Models/random_forest_model.pkl", "rb") as file:
    random_forest_model = pickle.load(file)

print("Models loaded successfully.")
```

```python
# Create test data

sample_data = pd.DataFrame({
    'rentEstimate_upperPrice': [2600],
    'floorAreaSqM': [45],
    'currentEnergyRating': [1],
    'bedrooms': [1],
    'bathrooms': [1],
    'livingRooms': [1],
    'tenure': [2],
    'propertyType': [12]
})

print("\nTest Data:")
print(sample_data)


# Make predictions

# Linear Regression Prediction
linear_prediction = linear_model.predict(sample_data)

# Random Forest Prediction
rf_prediction = random_forest_model.predict(sample_data)


# Display Results

print("\nLinear Regression Prediction:")
print(linear_prediction)

print("\nRandom Forest Prediction:")
print(rf_prediction)
```
